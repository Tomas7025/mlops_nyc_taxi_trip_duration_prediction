# Exploratory Data Analysis Notebook

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
import re

path_to_dir_data = '../data/raw/' # format: list of parquet files

def extract_data_from_parquet(file_path):
	df = pd.read_parquet(file_path)
	return df

list_of_files = os.listdir(path_to_dir_data)
list_of_files = [f for f in list_of_files if re.match(r'.*\.parquet$', f)]
print(f'Found {len(list_of_files)} files in {path_to_dir_data}')
print(list_of_files)
```

## Example of extracting data from the first parquet file

1. Data extraction from the first parquet file in the dataset directory and printing its general information.

```python

first_file_path = os.path.join(path_to_dir_data, list_of_files[0])
first_df = extract_data_from_parquet(first_file_path)

print(f'First file: {list_of_files[0]}')
print('General info about the first parquet file:')
print(first_df.info())
```
